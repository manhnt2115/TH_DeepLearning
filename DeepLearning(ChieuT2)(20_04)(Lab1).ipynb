```python
class GeometryModel(tf.Module):#13 14 15 16
    @tf.function
    def total_area_prism(self, a, b, c):
        # 15. Chu vi hình chữ nhật
        P = (a + b) * 2
        # 16. Diện tích toàn phần hình hộp
        S_xung_quanh = (a + b) * c * 2
        S_toan_phan = (a * b) * 2 + S_xung_quanh
        return P, S_toan_phan

model = GeometryModel()
p, s = model.total_area_prism(tf.constant(5.), tf.constant(3.), tf.constant(10.))
```

```python
@tf.function #17 18 19
def check_trapezoid(a, b, c, d):
    is_parallel = tf.equal(a/b, c/d) # Logic minh họa
    return is_parallel
```
